```python
def collatz(number):
  if number % 2 == 0:
    result = number // 2
    return result
  elif number % 2 == 1:
    result = 3 * number + 1
    return result
try:
  number = int(input('Enter an integer: '))
  while number != 1:
    number = Collatz(number)
    print(number, end=' ')
except ValueError:
  print('You did not enter an integer')
```
