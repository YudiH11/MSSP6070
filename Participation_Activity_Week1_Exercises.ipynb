## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean = sum(salaries)/len(salaries)
print(mean)
import statistics
print(statistics.mean(salaries))
```

### median

```python
middle = len(salaries)//2
median = (sorted(salaries)[middle-1]+sorted(salaries)[middle])//2
print(median)
print(statistics.median(salaries))
```

### mode

```python
mode = max(set(salaries),key=salaries.count)
print(mode)
print(statistics.mode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
mean = sum(salaries)/len(salaries)
squared = 0
for x in salaries:
    squared = squared+(x-mean)**2
sample_variance = squared/(len(salaries)-1)
print(sample_variance)
print(statistics.variance(salaries))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
mean = sum(salaries)/len(salaries)
squared = 0
for x in salaries:
    squared = squared+(x-mean)**2
sample_variance = squared/(len(salaries)-1)
sample_standard_deviation = sample_variance**0.5
print(sample_standard_deviation)
print(statistics.stdev(salaries))
```

## Exercise 6: Calculating more statistics
### range

### coefficient of variation

```python
coefficint_of_variation = sample_standard_deviation/mean
print(coefficint_of_variation)
```

### interquartile range

```python
sorted_salaries = sorted(salaries)
half_1 = sorted_salaries[:50]
half_2 = sorted_salaries[50:]
q1 = (half_1[24] + half_1[25]) / 2
q3 = (half_2[24] + half_2[25]) / 2
iqr = q3 - q1
print(iqr)
```

### quartile coefficent of dispersion

```python
sorted_salaries = sorted(salaries)
half_1 = sorted_salaries[:50]
half_2 = sorted_salaries[50:]
q1 = (half_1[24] + half_1[25]) / 2
q3 = (half_2[24] + half_2[25]) / 2
quartile_coefficient_of_dispersion = (q3-q1)/(q3+q1)
print(quartile_coefficient_of_dispersion)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_salary = min(salaries)
max_salary = max(salaries)
min_max_scaled = []
for x in salaries:
    min_max_scaled.append((x-min_salary)/(max_salary-min_salary))
print(min_max_scaled)
```

### standardizing

## Exercise 8: Calculating covariance and correlation
### covariance

```python
from statistics import mean, stdev
mean_salary = mean(salaries)
std_salary = stdev(salaries)
standardized = [(x - mean_salary) / std_salary for x in salaries]
standardized[:5]
import numpy as np
covariance = np.cov(min_max_scaled, standardized)
print(covariance)
```

### Pearson correlation coefficient ($\rho$)

```python
correlation = np.corrcoef(min_max_scaled, standardized)
print(correlation)
```
