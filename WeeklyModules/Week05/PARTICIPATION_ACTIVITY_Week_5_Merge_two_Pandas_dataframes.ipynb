```python
import pandas as pd
aapl = pd.read_csv('aapl.csv')
amzn = pd.read_csv('amzn.csv')
nflx = pd.read_csv('nflx.csv')
fb = pd.read_csv('fb.csv')
goog = pd.read_csv('goog.csv')
```

```python
# Ticker Column
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
nflx['ticker'] = 'NFLX'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
```

```python
faanf = pd.concat([aapl, amzn, nflx, fb, goog])
faanf.to_csv('faanf.csv', index=False)
```

```python
faanf['date'] = pd.to_datetime(faanf['date'])
faanf['volume'] = faanf['volume'].astype(int)
faanf = faanf.sort_values(['date', 'ticker'])
faanf.head()
```

```python
faanf.sort_values('volume', ascending=False).head(7)
```

```python
faanf_melted = faanf.melt(id_vars=['date', 'ticker'], var_name='variable', value_name='value')
faanf_melted.head()
```
