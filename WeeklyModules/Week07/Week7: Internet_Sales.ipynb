```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/MSSP6070/WeeklyModules/Week07/InternetSales.csv', encoding='cp1252', low_memory=False)
df.head()
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"])
df["Year"] = df["OrderDate"].dt.year
```

```python
yearly_sales = df.groupby("Year")["SalesAmount"].sum()
yearly_sales
```

```python
plt.figure(figsize=(10, 6))
plt.bar(yearly_sales.index, yearly_sales.values)
plt.title("Yearly Sales")
plt.xlabel("Year")
plt.ylabel("Sales Amount")
plt.savefig("Yearly_Internet_Sales.pdf")
plt.show()
```
