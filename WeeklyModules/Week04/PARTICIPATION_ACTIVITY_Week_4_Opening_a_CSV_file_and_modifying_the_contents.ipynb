```python
#2.D.
ring_locations = [
    'Bolivia',
    'Chile',
    'Ecuador',
    'Peru',
    'Costa Rica',
    'Guatemala',
    'Mexico',
    'Japan',
    'Philippines',
    'Indonesia',
    'New Zealand',
    'Antarctic',
    'Canada',
    'Fiji',
    'Alaska',
    'Washington',
    'California',
    'Russia',
    'Taiwan',
    'Tonga',
    'Kermadec Islands']
place = '|'.join(ring_locations)
df['parsed_place'].str.contains(place, na = False)
df['ring_of_fire'] = df['parsed_place'].str.contains(place, na = False)
df.loc[df['parsed_place'] == 'New Mexico', 'ring_of_fire'] = False
```
