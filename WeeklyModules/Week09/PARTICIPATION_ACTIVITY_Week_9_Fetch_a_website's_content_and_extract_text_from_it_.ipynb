```python
import requests
import os
import bs4
import time
```

```python
keyword = "cats"
max_downloads = 5

folder = "images"
os.makedirs(folder, exist_ok=True)
```

```python
url = "https://api.openverse.org/v1/images/"

params = {"q": "cats", "page_size": 5, "source": "flickr"}

res = requests.get(url, params=params, timeout=60)
res.raise_for_status()

data = res.json()

"Status code:", res.status_code
"Images found:", len(data["results"])
```

```python
data = res.json()
data
```

```python
images = data["results"]
"Number of images:", len(images)
```

```python
image_urls = []

for image in images:
    image_url = image.get("url")
    if image_url:
        image_urls.append(image_url)

image_urls
```

```python
from urllib.parse import urlparse

downloaded = 0

for i, image_url in enumerate(image_urls):
    try:
        print("Downloading:", image_url)

        image_res = requests.get(image_url, timeout=30)
        image_res.raise_for_status()

        filename = f"cat_{i+1}.jpg"
        filepath = os.path.join(folder, filename)

        with open(filepath, "wb") as image_file:
            for chunk in image_res.iter_content(100000):
                image_file.write(chunk)

        downloaded += 1
        print("Saved:", filepath)

    except requests.RequestException as e:
        print("Download failed:", e)

    time.sleep(2)

"Done!"
```
