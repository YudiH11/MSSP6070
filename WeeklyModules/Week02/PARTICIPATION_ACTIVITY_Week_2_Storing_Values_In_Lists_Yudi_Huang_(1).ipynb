```python
#3.
letters = ['a','b','c','d']
print(letters[int(int('3'*2)//11)])
# The answer is d. Because '3' is not int, is str, and this means repeat for two times,so '3'*2 is '33'
# Then int change str'33' to 33, and divided by 11 is 3
# Then find the third item in letters, cause python index from 0, so third item is d.
print(letters[-1])
# Thne answer is d. Because this is negative indexing.
# From left to right is -4, -3, -2, -1. So -1 is d.
print(letters[:2])
# Then answers are['a', 'b']. Because this is slicing, from start to end.
# So [:2] means from start to index 2.
```

```python
#5.
spam = ['apples', 'bananas', 'tofu', 'cats']
def commacode(items):
  result = ''
  for i in range(len(items)):
    if i == len(items) - 1:
      result += 'and ' + items[i]
    else:
      result += items[i] + ', '
  return result
print(commacode(spam))
```
